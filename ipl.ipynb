```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
```

```python
df=pd.read_csv('ipl_2026_matches.csv')
df
```

```python
print("rows: ",df.shape[0],"\ncolumns: ",df.shape[1])
```

```python
match_wins=df['winner'].value_counts()
sns.barplot(y=match_wins.index, x=match_wins.values,palette="rainbow")
plt.title("NO. OF MATCHES WON")
```

```python
sns.countplot(x=df['toss_decision'],palette="rainbow")
plt.title("TOSS DECISION TRENDS")
```

```python
count=df[df['toss_winner']==df['winner']]['match_no'].count()
per=(count*100)/df.shape[0]
per
```

```python

count=df['man_of_match'].value_counts().head(10)
sns.barplot(y=count.index,x=count.values,palette="rainbow")
plt.ylabel("PLAYER")
plt.xlabel("NO OF MATCHES")
plt.title("PLAYERS WITH MOST NO OF MAN OF THE MATCH AWARD")
```

```python
count=df['venue'].value_counts()
sns.barplot(y=count.index, x=count.values,palette="rainbow")
plt.xlabel("no of matches")
plt.ylabel("stadium")
plt.title("NO OF MATCHES IN DIFF STADIUMS")
```
