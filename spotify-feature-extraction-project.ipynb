```python
cols = ['popularity', 'duration_ms', 'danceability', 'energy', 'tempo', 'loudness']

for col in cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(col, ":", len(outliers))
```

```python
#top 10 most popular songs
top=df.sort_values(by='popularity',ascending=False)
top[['track_name','artists','track_genre','popularity']].head(10)
```

```python
#Har genre ki average popularity nikalo aur top 10 genres find karo
genre_popularity = df.groupby('track_genre')['popularity'].mean()
genre_popularity.sort_values(ascending=False).head(10)
```

```python
#Top 10 artists find karo jinke dataset mein sabse zyada tracks hain.
df1=df.groupby('artists')['track_id'].count()
df1.sort_values(ascending=False).head(10)
```

```python
#Top 10 genres find karo jinke tracks ki average danceability sabse zyada hai.
df1=df.groupby('track_genre')['danceability'].mean()
df1.sort_values(ascending=False).head(10)
```

```python
#Top 10 genres find karo jinke tracks ki average energy sabse zyada hai.
df1=df.groupby('track_genre')['energy'].mean()
df1.sort_values(ascending=False).head(10)
```

```python
#Explicit vs non-explicit tracks ka count nikalo.
df.groupby('explicit')['track_id'].count()
```

```python
#Har genre ka average energy aur average danceability ek saath nikalo.
df1=df.groupby('track_genre')[['energy','danceability']].mean()
df1
```

```python
#Top 10 tracks find karo jinka popularity high hai aur duration_min 5 minutes se zyada hai.
df[(df['popularity'] > 80) & (df['duration_min'] > 5)].sort_values(by='popularity', ascending=False).head(10)
```

```python
#Artists ke hisaab se average popularity nikalo, aur minimum 20 tracks wale artists mein se top 10 artists find karo.
artist_stats = df.groupby('artists').agg(
    track_count=('track_id', 'count'),
    avg_popularity=('popularity', 'mean')
)

artist_stats[artist_stats['track_count'] >= 20] \
    .sort_values('avg_popularity', ascending=False) \
    .head(10)
```

```python
#Har genre ka average popularity, energy, aur danceability nikalo. Phir avg_popularity ke basis par top 10 genres nikalo.
df1=df.groupby('track_genre').agg(
    avg_pop=('popularity','mean'),
    avg_energy=('energy','mean'),
    avg_dance=('danceability','mean'),
)
df1.sort_values(by='avg_pop',ascending=False).head(10)
```

```python
#Har track_genre ka sabse popular track find karo.
df1 = df.loc[df.groupby('track_genre')['popularity'].idxmax(),
             ['track_genre', 'track_name', 'artists', 'popularity']]

df1
```

```python
#Har genre ka average duration_min nikalo aur shortest average duration wale 10 genres find karo.
df1=df.groupby('track_genre')['duration_min'].mean()
df1.sort_values().head(10)
```

```python
# #popularity column ko NumPy array mein convert karo aur:

# Mean
# Median
# Standard deviation
# Minimum
# Maximum

# nikalo.


arr=np.array(df['popularity'])
arr
np.mean(arr)
np.median(arr)
np.std(arr)
np.min(arr)
np.max(arr)
```

```python
# popularity array arr mein:

# 25th percentile
# 50th percentile
# 75th percentile

# calculate karo using NumPy.
np.percentile(arr, [25, 50, 75])
```

```python
#popularity ko 0–1 range mein normalize karo using NumPy.
m1=np.min(arr)
m2=np.max(arr)
arr=(arr-m1)/(m2-m1)
arr
```

```python
#energy aur danceability ke NumPy arrays banao, phir correlation coefficient calculate karo.
arr1=np.array(df['energy'])
arr2= np.array(df['danceability'])
np.corrcoef(arr1,arr2)
```

```python
import matplotlib.pyplot as plt
```

```python
#popularity ka histogram banao using Matplotlib.

plt.hist(df['popularity'], bins=20)
plt.xlabel('Popularity')
plt.ylabel('Number of Tracks')
plt.title('Popularity Distribution')
plt.show()
```

```python
genre_popularity = df.groupby('track_genre')['popularity'].mean()
top10 = genre_popularity.sort_values(ascending=False).head(10)

plt.bar(top10.index, top10.values)
plt.xlabel('Genre')
plt.ylabel('Average Popularity')
plt.title('Top 10 Genres by Average Popularity')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.scatter(df['energy'], df['danceability'])
plt.xlabel('Energy')
plt.ylabel('Danceability')
plt.title('Energy vs Danceability')
plt.show()
```
